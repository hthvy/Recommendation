```python
from google.colab import drive
drive.mount('/content/gdrive/')
import pandas as pd
import numpy as np
```

```python
#Đọc dữ liệu từ file training
name = 'MovieLen100k'


#Đọc file training
training_df = pd.read_excel(r'/content/gdrive/MyDrive/Luận văn/{0}/{0}_training.xlsx'.format(name), dtype={'User': int, 'Item': int})

#Tính trung bình rating của user
user = []
user_average = []
n_user = max(training_df.User)

for i in range(1, n_user + 1):
  rate_of_user = training_df.Rating[training_df.User == i].to_list()
  try:
    aver = sum(rate_of_user)/len(rate_of_user)
    user.append(i)
    user_average.append(aver)
  except:
    1

#Tạo data frame chứa thông tin, xuất ra file excel
df = pd.DataFrame(data={'User':user, 'User_Average': user_average})
df.to_excel("/content/gdrive/MyDrive/Luận văn/{0}/User_Average.xlsx".format(name), index=False)


#Tính trung bình rating của Item
Item = []
Item_Average = []
n_item = np.max(training_df.Item)

for i in range(1, n_item + 1):
  items = training_df.Rating[training_df.Item == i].to_list()
  try:
    avg = sum(items)/len(items)
    Item.append(i)
    Item_Average.append(avg)
  except:
    1

#Tạo data frame chứa thông tin, xuất ra file excel
df = pd.DataFrame(data={'Item':Item, 'Item_Average': Item_Average})
df.to_excel("/content/gdrive/MyDrive/Luận văn/{0}/Item_Average.xlsx".format(name), index=False)
```
