```python
#đọc file training, user_average
name = 'MovieLen100k'


#Đọc file training, user_average, item_average
training_df = pd.read_excel(r'/content/gdrive/MyDrive/Luận văn/{0}/{0}_training.xlsx'.format(name), dtype={'User': int, 'Item': int})
user_average_df = pd.read_excel(r'/content/gdrive/MyDrive/Luận văn/{0}/User_Average.xlsx'.format(name), dtype={'User': int})
item_average_df = pd.read_excel(r'/content/gdrive/MyDrive/Luận văn/{0}/Item_Average.xlsx'.format(name), dtype={'Item': int})
```

```python
#Tính sim giữa 2 user
def calc_sim_2item(TrainingDataFrame, UserAverageDataFrame, Item1, Item2):
    #Lấy danh sách user đã đánh giá trên từng item
    Item1_user = TrainingDataFrame.User[TrainingDataFrame.Item == Item1][TrainingDataFrame.Rating >= 0].to_list()
    Item2_user = TrainingDataFrame.User[TrainingDataFrame.Item == Item2][TrainingDataFrame.Rating >= 0].to_list()

    #Tập user đã đánh giá trên 2 item
    intersection = [value for value in Item1_user if value in Item2_user]

    #Hợp của 2 tập item
    union = Item1_user + Item2_user
    union = list(set(union))

    #Chặn lại trong trường hợp không có item chung hay là mẫu bằng 0
    if len(intersection) == 0:
        return 0
    if (len(Item1_user) + len(Item2_user) - len(intersection)) == 0:
        return 0

    #Tính tử số của biểu thức sim
    set1 = []
    set2 = []
    for i in intersection:
        if(abs((((TrainingDataFrame.Rating[TrainingDataFrame.User == i][TrainingDataFrame.Item == Item1].values[0]) - (UserAverageDataFrame.User_Average[UserAverageDataFrame.User == i].values[0]))
        - ((TrainingDataFrame.Rating[TrainingDataFrame.User == i][TrainingDataFrame.Item == Item2].values[0]) - (UserAverageDataFrame.User_Average[UserAverageDataFrame.User == i].values[0])))) <= 0.5):
            set1.append(i)
        elif(0.5 < abs((((TrainingDataFrame.Rating[TrainingDataFrame.User == i][TrainingDataFrame.Item == Item1].values[0]) - (UserAverageDataFrame.User_Average[UserAverageDataFrame.User == i].values[0]))
        - ((TrainingDataFrame.Rating[TrainingDataFrame.User == i][TrainingDataFrame.Item == Item2].values[0]) - (UserAverageDataFrame.User_Average[UserAverageDataFrame.User == i].values[0])))) <= 1):
            set2.append(i)


    return ((len(set1) * 1 + len(set2) * 0.5) / len(union))
```

```python
#tính sim 1 item với tất cả item khác
def sim_idf(TrainingDataFrame, User_Average_DataFrame, Item_Average_DataFrame, Item):
    Item_col =[]
    Sim_col = []
    print('Sim của Item: ', Item)

    #Vòng lặp hết item
    for i in Item_Average_DataFrame.Item.to_list():
      if Item != i:
          Item_col.append(i)
          Sim_col.append(calc_sim_2item(TrainingDataFrame, User_Average_DataFrame, Item, i))

    #Tạo dataframe
    df = pd.DataFrame(data={'Item':Item_col, 'Sim':Sim_col})

    #Sắp xếp độ tương đồng
    df = df.sort_values(by=['Sim'], axis=0,ascending=False, ignore_index=True)
    df.to_excel('/content/gdrive/MyDrive/Luận văn/{0}/SimItem/Item{1}.xlsx'.format(name, Item), index=False)
```

```python
#Tính sim toàn bộ item
def to_Excel(TrainingDataFrame, User_Average_DataFrame, Item_Average_DataFrame):
  for i in Item_Average_DataFrame.Item.to_list():
    sim_idf(TrainingDataFrame, User_Average_DataFrame, Item_Average_DataFrame, i)


to_Excel(training_df, user_average_df, item_average_df)
```
