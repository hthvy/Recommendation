```python
#đọc file training, user_average
name = 'Movielen100k'

#Đọc file training, user_average
training_df = pd.read_excel(r'/content/gdrive/MyDrive/Luận văn/{0}/{0}_training.xlsx'.format(name), dtype={'User': int, 'Item': int})
user_average_df = pd.read_excel(r'/content/gdrive/MyDrive/Luận văn/{0}/User_Average.xlsx'.format(name), dtype={'User': int})
```

```python
#Tính sim giữa 2 user
def calc_sim_2user(TrainingDataFrame, UserAverageDataFrame, User1, User2):
    #Lấy danh sách item của từng user
    User1_items = TrainingDataFrame.Item[TrainingDataFrame.User == User1][TrainingDataFrame.Rating >= 0].to_list()
    User2_items = TrainingDataFrame.Item[TrainingDataFrame.User == User2][TrainingDataFrame.Rating >= 0].to_list()

    #Tập item chung được đánh giá bởi 2 user
    intersection = [value for value in User1_items if value in User2_items]

    #Hợp của 2 người dùng
    union = User1_items + User2_items
    union = list(set(union))

    #Chặn lại trong trường hợp không có item chung hay là mẫu bằng 0
    if len(intersection) == 0:
        return 0
    if (len(User1_items) + len(User2_items) - len(intersection)) == 0:
        return 0

    #Tính tử số của biểu thức sim
    set1 = []
    set2 = []
    for i in intersection:
        if(abs((((TrainingDataFrame.Rating[TrainingDataFrame.User == User1][TrainingDataFrame.Item == i].values[0]) - (UserAverageDataFrame.User_Average[UserAverageDataFrame.User == User1].values[0]))
        - ((TrainingDataFrame.Rating[TrainingDataFrame.User == User2][TrainingDataFrame.Item == i].values[0]) - (UserAverageDataFrame.User_Average[UserAverageDataFrame.User == User2].values[0])))) <= 0.5):
            set1.append(i)
        elif(0.5 < abs((((TrainingDataFrame.Rating[TrainingDataFrame.User == User1][TrainingDataFrame.Item == i].values[0]) - (UserAverageDataFrame.User_Average[UserAverageDataFrame.User == User1].values[0]))
        - ((TrainingDataFrame.Rating[TrainingDataFrame.User == User2][TrainingDataFrame.Item == i].values[0]) - (UserAverageDataFrame.User_Average[UserAverageDataFrame.User == User2].values[0])))) <= 1):
            set2.append(i)


    return ((len(set1) * 1 + len(set2) * 0.5) / len(union))
```

```python
#tính sim 1 user với tất cả user khác
def sim_df(TrainingDataFrame, User_Average_DataFrame, User):
    User_col =[]
    Sim_col = []
    print('Sim của user: ', User)

    #Vòng lặp hết người dùng
    for i in User_Average_DataFrame.User.to_list():
      if User != i:
          User_col.append(i)
          Sim_col.append(calc_sim_2user(TrainingDataFrame, User_Average_DataFrame, User, i))

    #Tạo dataframe
    df = pd.DataFrame(data={'User':User_col, 'Sim':Sim_col})

    #Sắp xếp độ tương đồng
    df = df.sort_values(by=['Sim'], axis=0,ascending=False, ignore_index=True)
    df.to_excel('/content/gdrive/MyDrive/Luận văn/{0}/SimUser/User{1}.xlsx'.format(name, User), index=False)
```

```python
#Tính sim toàn bộ người dùng
def to_Excel(TrainingDataFrame, User_Average_DataFrame):
  for i in User_Average_DataFrame.User.to_list():
    sim_df(TrainingDataFrame, User_Average_DataFrame, i)


to_Excel(training_df, user_average_df)
```
